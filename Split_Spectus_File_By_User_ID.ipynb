```python
import csv
import os

print(os.getcwd())

base_path = '../Mode_Analysis_Big_Data/20_minute_stops'
directory_path = f'{base_path}/split_by_user'
os.makedirs(directory_path, exist_ok=True)

with open(f"{base_path}/20260820_seattle_2000_processed_20min_stop_200m_radius.csv", "r", newline="") as infile:
    reader = csv.reader(infile)
    header = next(reader)
    user_id_index = header.index("user_ID")  # auto-detect column position

    current_user = None
    count = 0
    outfile = None
    writer = None

    for row in reader:
        user_id = row[user_id_index]

        if user_id != current_user:
            if outfile:
                outfile.close()
                count += 1
                print(f'Processed user {current_user}')
                print(f'{count} total users processed.')
            current_user = user_id
            outfile = open(f"{directory_path}/user_{user_id}.csv", "w", newline="")
            writer = csv.writer(outfile)
            writer.writerow(header)

        writer.writerow(row)

    if outfile:
        outfile.close()
        count += 1
        print(f'Processed user {current_user}')
        print(f'{count} total users processed.')
```
