```python
import csv
import os

os.makedirs("./data/Spectus/Lyra_Processed/split_by_user", exist_ok=True)

with open("./data/Spectus/Lyra_Processed/Seattle_2000_processed_data.csv", "r", newline="") as infile:
    reader = csv.reader(infile)
    header = next(reader)
    user_id_index = header.index("user_ID")  # auto-detect column position

    current_user = None
    count = 0
    outfile = None
    writer = None

    for row in reader:
        user_id = row[user_id_index]

        if user_id != current_user:
            if outfile:
                outfile.close()
                count += 1
                print(f'Processed user {current_user}')
                print(f'{count} total users processed.')
            current_user = user_id
            outfile = open(f"./data/Spectus/Lyra_Processed/split_by_user/user_{user_id}.csv", "w", newline="")
            writer = csv.writer(outfile)
            writer.writerow(header)

        writer.writerow(row)

    if outfile:
        outfile.close()
        count += 1
        print(f'Processed user {current_user}')
        print(f'{count} total users processed.')
```
