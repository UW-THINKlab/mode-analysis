```python
import matplotlib.pyplot as plt
import pandas as pd

df_spectus = pd.read_csv(f'../data/Spectus/Lyra_Processed/split_by_user/database_metrics.csv')

filtered = df_spectus[df_spectus['number_of_records'] <= 400]
num_filtered = len(df_spectus[df_spectus['number_of_records'] > 400])
print(f'Filtered {num_filtered} records ({num_filtered / len(df_spectus) * 100.00}%)')
value_counts = filtered['number_of_records'].value_counts().sort_index()

median_val = filtered['number_of_records'].median()
p90_val = filtered['number_of_records'].quantile(0.90)

plt.figure(figsize=(10, 6))
plt.hist(filtered['number_of_records'], bins=40, range=(0, 400), color= 'red', edgecolor='black')

plt.axvline(median_val, color='black', linestyle='--', linewidth=2, label=f'Median: {median_val:.0f}')
plt.axvline(p90_val, color='orange', linestyle='--', linewidth=2, label=f'P90: {p90_val:.0f}')
plt.legend()

plt.xlabel('Number of Records per Trip')
plt.ylabel('Number of Trips')
plt.title('Distribution of Number of Records per Trip')
plt.show()
```
