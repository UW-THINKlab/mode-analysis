```python
import os
from src.geospatial_analysis import evaluate_transit_likelihood
from src.optimized_analysis import get_mode_from_file_path
from src.plot import plot_trip_on_map
from src.process_files_rmove import get_dataframe_from_file_path

max_count = 200
root_path = "../data/rMove/Processed_Trajectory/"

count = 0
is_transit_count = 0
for f in os.scandir(root_path):
    # if "Transit" in f.name:
    if "Transit" not in f.name:
        continue
    if max_count == None or count < max_count:
        print(f"---- Processing {f.name} ----")
        mode = get_mode_from_file_path(f.path)
        print(f"Mode: {mode}")
        df = get_dataframe_from_file_path(f.path)
        is_transit = evaluate_transit_likelihood(df)
        print(f"Is transit: {is_transit}")
        if is_transit: is_transit_count += 1
        # plot_trip_on_map(df, include_transit=True)
        count += 1

print("---------------------")
print(f"Total Count: {count}")
print(f"Is transit count: {is_transit_count}")
```
