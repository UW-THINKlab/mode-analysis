```python
import matplotlib.pyplot as plt
import pandas as pd

ax = axes_flat[2]

root_path = '../data/'
file_name = 'database_metrics.csv'

df_geolife = pd.read_csv(f'{root_path}Geolife/{file_name}')
df_geolife = df_geolife[df_geolife['number_of_records'] > 0]

filtered = df_geolife[df_geolife['number_of_records'] <= 4000]
num_filtered = len(df_geolife[df_geolife['number_of_records'] > 4000])
print(f'Filtered {num_filtered} records ({num_filtered / len(df_geolife) * 100.00}%)')
value_counts = filtered['number_of_records'].value_counts().sort_index()

median_val = filtered['number_of_records'].median()
p90_val = filtered['number_of_records'].quantile(0.90)

# plt.figure(figsize=(10, 6))
ax.hist(filtered['number_of_records'], bins=40, range=(0, 4000), color= 'red', edgecolor='black')

ax.axvline(median_val, color='black', linestyle='--', linewidth=2, label=f'Median: {median_val:.0f}')
ax.axvline(p90_val, color='orange', linestyle='--', linewidth=2, label=f'P90: {p90_val:.0f}')
ax.legend()

ax.set_xlabel('Number of Records per Trip')
ax.set_ylabel('Number of Trips')
ax.set_title('Distribution of Number of Records per Trip')
# plt.show()

fig
```
