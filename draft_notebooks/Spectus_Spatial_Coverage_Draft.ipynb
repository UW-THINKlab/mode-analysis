```python
import contextily as ctx
import geopandas as gpd
import matplotlib.pyplot as plt
import os
import pandas as pd
from shapely.geometry import LineString

# directory = './data/Spectus/Lyra_Processed/split_by_user'

df_spectus = pd.read_csv(f'../data/Spectus/Lyra_Processed/Seattle_2000_compressed_trips.csv')

print(len(df_spectus))

def plot_trips_on_map(start_index=None,
                      end_index=None,
                      base_path='data/Spectus/Lyra_Processed/split_by_user/user_EXAMPLE.csv'):

    points_dataframe = pd.read_csv(base_path)
    points_dataframe = points_dataframe[points_dataframe['traj_id'] != -99]
    points_dataframe = points_dataframe.rename(columns={'orig_lat': 'lat', 'orig_long': 'lng', 'datetime': 'timestamp'})
    points_dataframe['timestamp'] = pd.to_datetime(points_dataframe['timestamp'])

    points_geodataframe = gpd.GeoDataFrame(points_dataframe,
                                           geometry=gpd.points_from_xy(points_dataframe.lng,
                                             points_dataframe.lat))
    points_geodataframe.crs = "EPSG:4326"  # For lat/lon data
    points_geodataframe = points_geodataframe.to_crs(epsg=3857)

    fig, ax = plt.subplots(dpi=200)

    minx, miny, maxx, maxy = points_geodataframe.total_bounds
    ax.set_xlim(minx, maxx)
    ax.set_ylim(miny, maxy)

    count = 0
    for value, group in points_geodataframe.groupby('traj_id'):
        if end_index != None and count >= end_index:
            break
        if start_index != None and count < start_index:
            count += 1
            continue

        print(f'Processing Trajectory {value}')

        line = LineString(points_geodataframe.geometry.tolist())
        line_gdf = gpd.GeoDataFrame(geometry=[line], crs=points_geodataframe.crs)
        line_gdf.plot(ax=ax, color="black", linewidth=0.1, zorder=3)

        count += 1

    ctx.add_basemap(ax, source=ctx.providers.OpenStreetMap.Mapnik)

    plt.show()

plot_trips_on_map(start_index=0, end_index=306)
```

```python
import geopandas as gpd

cbg_gdf = gpd.read_file("../data/Seattle_neighborhoods/nma_nhoods_main.shp")
cbg_gdf = cbg_gdf.to_crs("EPSG:4326")  # match your points' CRS

# print(cbg_gdf.columns.tolist())

# start_index = 0
# end_index = 10000
points_dataframe = pd.read_csv('../data/Spectus/Lyra_Processed/Seattle_2000_compressed_trips.csv')
points_dataframe = points_dataframe[points_dataframe['traj_id'] != -99]
# points_dataframe = points_dataframe.iloc[start_index:end_index]
points_dataframe = points_dataframe.rename(columns={'start_lat': 'lat', 'start_lon': 'lng'})

points_geodataframe = gpd.GeoDataFrame(points_dataframe,
                                       geometry=gpd.points_from_xy(points_dataframe.lng,
                                         points_dataframe.lat))
points_geodataframe.crs = "EPSG:4326"  # For lat/lon data
joined = gpd.sjoin(points_geodataframe, cbg_gdf, how="left", predicate="within")

# count how many points fell in each CBG
counts = joined.groupby("L_HOOD").size().rename("point_count")  # GEOID or whatever your CBG id column is called

cbg_gdf = cbg_gdf.merge(counts, on="L_HOOD", how="left")
cbg_gdf["point_count"] = cbg_gdf["point_count"].fillna(0)

fig, ax = plt.subplots()

cbg_gdf.plot(
    column="point_count",
    cmap="Blues",           # light = low, dark = high
    edgecolor="black",
    linewidth=0.3,
    ax=ax,
    zorder=3,
    legend=True,
)

# ctx.add_basemap(ax, crs="EPSG:4326", source=ctx.providers.OpenStreetMap.Mapnik)

plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.set_title("Starting Location by Neighborhood")
plt.show()
```
