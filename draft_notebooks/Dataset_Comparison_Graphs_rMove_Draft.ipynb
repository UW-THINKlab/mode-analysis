```python
import matplotlib.pyplot as plt
import pandas as pd

root_path = '../data/'
file_name = 'database_metrics.csv'

df_rmove = pd.read_csv(f'{root_path}rMove/{file_name}')
# df_spectus = pd.read_csv(f'./data/Spectus/spectus_database_metrics.csv')

filtered = df_rmove[df_rmove['number_of_records'] <= 200]
value_counts = filtered['number_of_records'].value_counts().sort_index()

median_val = filtered['number_of_records'].median()
p90_val = filtered['number_of_records'].quantile(0.90)

plt.figure(figsize=(10, 6))
plt.hist(filtered['number_of_records'], bins=40, range=(0, 200), edgecolor='black')

plt.axvline(median_val, color='red', linestyle='--', linewidth=2, label=f'Median: {median_val:.0f}')
plt.axvline(p90_val, color='orange', linestyle='--', linewidth=2, label=f'P90: {p90_val:.0f}')
plt.legend()

plt.xlabel('Number of Records per Trip')
plt.ylabel('Number of Trips')
plt.title('Distribution of Number of Records per Trip')
plt.show()
```
