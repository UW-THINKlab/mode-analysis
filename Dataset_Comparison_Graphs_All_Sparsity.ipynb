# Sparsity across all Datasets

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, PercentFormatter

fig, axes = plt.subplots(4, 3, figsize=(18, 16))

col_titles = ["GeoLife", "rMove", "Spectus"]
row_titles = ["5 Second Intervals", "20 Second Intervals", "30 Second Intervals", "60 Second Intervals"]

# Column titles: set on top row only
for ax, col in zip(axes[0], col_titles):
    ax.annotate(col, xy=(0.5, 1), xytext=(0, 30),
            xycoords='axes fraction', textcoords='offset points',
            fontsize=16, fontweight='bold', ha='center', va='baseline')

# Row titles: place to the left of each row's first axis
for ax, row in zip(axes[:, 0], row_titles):
    ax.annotate(row, xy=(0, 0.5), xytext=(-ax.yaxis.labelpad - 5, 0),
                xycoords=ax.yaxis.label, textcoords='offset points',
                fontsize=16, fontweight='bold', ha='right', va='center', rotation=90)

axes_flat = axes.flat
```

```python
import matplotlib.pyplot as plt
import pandas as pd

ax = axes_flat[0]

root_path = 'data/'
file_name = '20260714_geolife_database_metrics.csv'

df_geolife = pd.read_csv(f'{root_path}Geolife/{file_name}')
num_rows_original = len(df_geolife)
df_geolife = df_geolife[df_geolife['number_of_records'] > 0]
num_rows_after = len(df_geolife)
print(f'Originally: {num_rows_original}, Afterward: {num_rows_after}, Removed: {(num_rows_original - num_rows_after)/num_rows_original * 100: .2f}%')

filtered = df_geolife
value_counts = filtered['sparsity_5s'].value_counts().sort_index()

p10_val = filtered['sparsity_5s'].quantile(0.10)
median_val = filtered['sparsity_5s'].median()
p90_val = filtered['sparsity_5s'].quantile(0.90)

# plt.figure(figsize=(10, 6))
ax.hist(filtered['sparsity_5s'], bins=50, range=(0.0, 1.0), color='lightgrey', edgecolor='black')

ax.axvline(p10_val, color='lime', linestyle='--', linewidth=2, label=f'10th Percentile: {p10_val:.2f}')
ax.axvline(median_val, color='red', linestyle='--', linewidth=2, label=f'Median: {median_val:.2f}')
ax.axvline(p90_val, color='orange', linestyle='--', linewidth=2, label=f'90th Percentile: {p90_val:.2f}')
ax.legend()

total_count = len(filtered['sparsity_5s'])
ax.yaxis.set_major_formatter(PercentFormatter(xmax=total_count, decimals=0))
ax.yaxis.set_major_locator(MultipleLocator(total_count * 0.10))
ax.set_ylim(0, total_count/3)
ax.set_xlabel('Sparsity of Trip (% of intervals containing datapoints)')
ax.set_ylabel('Number of Trips')
ax.set_title('GeoLife - Distribution of Trip Sparsity (5s intervals)')
# plt.show()

fig
```
