# Sparsity across all Datasets

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(3, 4, figsize=(24, 12))
axes_flat = axes.flat
```

```python
import matplotlib.pyplot as plt
import pandas as pd

ax = axes_flat[0]

root_path = 'data/'
file_name = '20260714_geolife_database_metrics.csv'

df_geolife = pd.read_csv(f'{root_path}Geolife/{file_name}')
num_rows_original = len(df_geolife)
df_geolife = df_geolife[df_geolife['number_of_records'] > 0]
num_rows_after = len(df_geolife)
print(f'Originally: {num_rows_original}, Afterward: {num_rows_after}, Removed: {(num_rows_original - num_rows_after)/num_rows_original * 100: .2f}%')

filtered = df_geolife
value_counts = filtered['sparsity_5s'].value_counts().sort_index()

p10_val = filtered['sparsity_5s'].quantile(0.10)
median_val = filtered['sparsity_5s'].median()
p90_val = filtered['sparsity_5s'].quantile(0.90)

# plt.figure(figsize=(10, 6))
ax.hist(filtered['sparsity_5s'], bins=50, range=(0.0, 1.0), color='grey', edgecolor='black')

ax.axvline(p10_val, color='lime', linestyle='--', linewidth=2, label=f'10th Percentile: {p10_val:.2f}')
ax.axvline(median_val, color='red', linestyle='--', linewidth=2, label=f'Median: {median_val:.2f}')
ax.axvline(p90_val, color='orange', linestyle='--', linewidth=2, label=f'90th Percentile: {p90_val:.2f}')
ax.legend()

ax.yaxis.set_major_formatter('{x:,.0f}')
ax.set_xlabel('Sparsity of Trip (% of intervals containing datapoints)')
ax.set_ylabel('Number of Trips')
ax.set_title('GeoLife - Distribution of Trip Sparsity (5s intervals)')
# plt.show()

fig
```
